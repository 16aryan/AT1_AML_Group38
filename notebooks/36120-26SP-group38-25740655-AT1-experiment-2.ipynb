## B. Experiment Description

```python
experiment_hypothesis = """
Experiment 2 keeps the player-level representation established in Experiment 1 and changes only the algorithm, replacing Logistic Regression with a Random Forest. The hypothesis is that a bagged tree ensemble will rank players more effectively than a linear model, because the relationship between college statistics and professional longevity is unlikely to be additive: a high usage rate means something very different for an efficient scorer than for an inefficient one, and a linear model cannot represent that without the interaction being specified by hand.

A second question is tested alongside it. Experiment 1 fixed the aggregation set without justification, so this experiment ablates it, comparing the latest-plus-mean-plus-max set against a fuller set that also carries the minimum and standard deviation of every statistic.
"""
```

## I. Selection of Performance Metrics

> Provide some explanations on why you believe the performance metrics you chose is appropriate

```python
# This table is the argument for the metric choice. It scores three deliberately
# useless models against the validation labels; a metric that rewards them
# cannot be trusted for this problem.
rng = np.random.default_rng(RANDOM_STATE)
degenerate = {
    "always predicts 0": np.zeros(len(y_val)),
    "always predicts 1": np.ones(len(y_val)),
    "uniform random": rng.random(len(y_val)),
}
rows = []
for name, scores in degenerate.items():
    preds = (scores >= 0.5).astype(int)
    rows.append({"model": name,
                 "accuracy": (preds == y_val).mean(),
                 "roc_auc": roc_auc_score(y_val, scores) if len(np.unique(scores)) > 1 else np.nan,
                 "auprc": average_precision_score(y_val, scores)})
display(pd.DataFrame(rows).round(4))
print("Positive class rate (the AUPRC baseline):", round(y_val.mean(), 4))
```

```python
performance_metrics_explanations = """
AUPRC, reported as average precision, is the primary metric because it is what the competition scores and because the target is severely imbalanced at roughly 3.4% positive.

The comparison above shows why the alternatives fail. A model predicting the negative class for every player achieves about 96.6% accuracy while identifying no future NBA players at all, which disqualifies accuracy outright. ROC-AUC is more subtle but still misleading: it is computed from true-positive and false-positive rates, and because the negative class is roughly twenty-eight times larger, a large absolute number of false positives moves the false-positive rate only slightly. A model can therefore post a respectable ROC-AUC while the players it ranks highest are overwhelmingly wrong, which is precisely the failure a scouting department would experience. AUPRC uses precision in place of the false-positive rate and so responds directly to that failure.

ROC-AUC is still reported as a secondary metric for comparability across the three experiments, and precision, recall and F1 are examined across thresholds in Section J.4. Lift over the AUPRC baseline is reported too, because a raw AUPRC value means very different things at different base rates.
"""
```

```python
# Ablation: does the fuller aggregation set earn its extra 116 columns?
alt_train = build_player_table(train_season, numeric_columns,
                               stats=("latest", "mean", "max", "min", "std"),
                               include_target=True)
alt_test = build_player_table(test_season, numeric_columns,
                              stats=("latest", "mean", "max", "min", "std"))
X_alt = alt_train.drop(columns=["pid", target_name])
X_alt_kaggle = alt_test.drop(columns=["pid"])
X_alt, _ = align_categoricals(X_alt, X_alt_kaggle)

rows = []
for name, frame in [("latest+mean+max (181 feats)", X_all),
                    ("latest+mean+max+min+std", X_alt)]:
    oof, _ = out_of_fold_predictions(
        lambda: RankScorer(build_random_forest()), frame, y_all, n_splits=5)
    rows.append({"aggregation_set": name, "features": frame.shape[1],
                 "oof_auprc": round(average_precision_score(y_all, oof), 4)})
display(pd.DataFrame(rows))
```

```python
# Minimum leaf size controls how finely the forest may partition a rare class.
rows = []
for leaf in (1, 5, 20):
    oof, _ = out_of_fold_predictions(
        lambda leaf=leaf: RankScorer(build_random_forest(min_samples_leaf=leaf)),
        X_all, y_all, n_splits=5)
    rows.append({"min_samples_leaf": leaf,
                 "oof_auprc": round(average_precision_score(y_all, oof), 4)})
display(pd.DataFrame(rows))
best_leaf = max(rows, key=lambda r: r["oof_auprc"])["min_samples_leaf"]
print("Selected min_samples_leaf:", best_leaf)
```

```python
hyperparameters_selection_explanations = """
Three settings were chosen on evidence and the rest left at values that are standard for an imbalanced problem.

The aggregation set was ablated first, because it changes the feature matrix and therefore conditions every later choice. The smaller set was retained, confirming the expectation that roughly 600 positive players cannot support 297 columns.

Minimum leaf size was then searched over a small grid. It is the parameter that matters most here: a forest permitted to isolate single players will construct leaves containing one positive case each, which is memorisation rather than signal at this base rate.

The number of trees is set to 600 and not tuned, because forest performance is monotonic and flattening in that parameter, so more trees never hurt beyond the compute they cost. Feature subsampling is left at the square root of the feature count, which is the standard choice for decorrelating trees. Class weighting uses 'balanced_subsample', which reweights by inverse class frequency within each bootstrap sample rather than once globally, which is the appropriate variant when the resampling itself changes the class balance seen by each tree.

The same caveat applies as in Experiment 1: these folds are drawn from within the 2008-2016 training years, so they measure same-era ranking. Experiment 3 quantifies how much that flatters a result.
"""
```

### J.5 Business Impact from Current Model Performance

> Provide some analysis on the model impacts from the business point of view

```python
# The threshold sweep is a technical artefact until expressed as scouting
# workload: given capacity to evaluate N players in depth, how many genuine
# future NBA players does that shortlist contain?
rows = []
total_positives = int(y_val.sum())
for size in (10, 25, 50, 100, 200):
    if size > len(y_val):
        continue
    cutoff = np.sort(val_probabilities)[-size]
    picked = val_probabilities >= cutoff
    found = int(y_val[picked].sum())
    rows.append({"shortlist_size": size,
                 "genuine_prospects_found": found,
                 "precision_%": round(found / picked.sum() * 100, 1),
                 "recall_%": round(found / total_positives * 100, 1),
                 "lift_vs_random": round((found / picked.sum()) / y_val.mean(), 1)})
print(f"Validation pool: {len(y_val)} players, {total_positives} reached a "
      f"sustained NBA career ({y_val.mean():.2%}).")
display(pd.DataFrame(rows))
```

```python
business_impacts_explanations = """
The model's value is prioritisation rather than classification. The shortlist table reframes the threshold choice as the question a scouting department faces: given capacity to evaluate a fixed number of players in depth, how many genuine future NBA players does that shortlist contain and how many are missed?

The lift column states how much better the shortlist is than drawing the same number of players at random, and it is the figure that translates most directly into operational value, because evaluation capacity is always far smaller than the candidate pool.

Two limitations bound practical use. Because the positive class is so rare, recall at a workable shortlist size is inevitably limited: the model narrows a search rather than closing it, and false negatives represent players with genuine NBA futures who never receive attention. The threshold must come from the organisation's own costs and capacity, not from validation F1, which is reported only as a diagnostic. A specific fairness risk also applies to this model: imputing missing statistics with a median systematically affects players who played fewer minutes, who are disproportionately from smaller programmes, so subgroup performance by conference should be audited before deployment.
"""
```
